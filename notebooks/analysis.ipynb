## Key Findings & Recommendations

The committed figures use a synthetic demo, so treat these as a decision framework until the Olist data is loaded.

- Compare revenue with AOV each period; divergent movement signals whether growth comes from order volume or basket expansion.
- Focus merchandising and seller-quality investigation on high-revenue categories, while measuring whether their share is becoming over-concentrated.
- Use cohort decay to prioritize post-delivery cross-sell and replenishment journeys where retention drops.
- Use customer order sequence and running revenue to define repeat-buyer segments and study which first purchases lead to a second order.
